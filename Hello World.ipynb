```python
import pandas as pd
# Example: Replace this with the raw URL of your GitHub file
github_raw_url = 'https://raw.githubusercontent.com/teng71/ADALL/refs/heads/main/laptop_prices_2024_sgd_TL.csv'
try:
    df = pd.read_csv(github_raw_url)
    print("Successfully loaded data from GitHub!")
    display(df.head())

    #Import plt library
    import matplotlib.pyplot as plt

    #Plot each df col graph
    df.plot(kind='line')
    plt.show()

    #Generate histogram
    df.hist()
    plt.show()

except Exception as e:
    print(f"Error loading data: {e}")
    print("Please ensure the URL is correct and the file format is compatible with `pd.read_csv`.")
```
